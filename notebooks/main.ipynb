```python
OVERWRITE = False
MIN_SIZE = int(1e3)
MAX_SIZE = int(1e5)
SUBSET = 32
DOWNSCALE = 6
BATCH_SIZE = 1024
BLOCK_SIZE = (2 ** 8) * (2 ** 12)
SPECIALS = {
    "unk_token": "<unk>",
    "pad_token": "<pad>",
    "mask_token": "<mask>",
    "bos_token": "<bos>",
    "eos_token": "<eos>",
    "cls_token": "<cls>",
}
VOCAB_SIZE = (2 ** 8) * (2 ** 6)
```

```python
BYTE_TO_UTF8 = {i: chr(i + 10752) for i in range(256)}
```

```python
p = Path("/home/lk3591/Documents/datasets/Sorel/processed/")
sorel_files = list(chain((p / "train").iterdir(), (p / "test").iterdir()))
if len(sorel_files) > SUBSET:
    sorel_files = sorel_files[0:SUBSET]

p = Path("/home/lk3591/Documents/datasets/Windows/processed/")
windows_files = list(chain((p / "train").iterdir(), (p / "test").iterdir()))
if len(windows_files) > SUBSET:
    windows_files = windows_files[0:SUBSET]

def byte_string(p: Path) -> str:
    return "".join([BYTE_TO_UTF8[b] for b in p.read_bytes()])

def dataset_generator():
    for p in sorel_files:
        if MIN_SIZE <= p.stat().st_size <= MAX_SIZE:
            yield {"text": byte_string(p), "label": 1, "file": p.as_posix()}
    for p in windows_files:
        if MIN_SIZE <= p.stat().st_size <= MAX_SIZE:
            yield {"text": byte_string(p), "label": 0, "file": p.as_posix()}
```

```python
dataset = Dataset.from_generator(dataset_generator)
```

```python
def group_texts(examples):
    concatenated = "".join(examples["text"])
    total_length = len(concatenated)
    if total_length >= BLOCK_SIZE:
        total_length = (total_length // BLOCK_SIZE) * BLOCK_SIZE
    chopped = [concatenated[i : i + BLOCK_SIZE] for i in range(0, total_length, BLOCK_SIZE)]
    examples["text"] = chopped
    return examples

grouped_dataset = dataset.remove_columns(["label", "file"]).map(group_texts, batched=True)
assert len(grouped_dataset[0]["text"]) == BLOCK_SIZE
grouped_dataset[0]["text"]
```

```python
def tokenization_generator(batch_size: int = BATCH_SIZE):
    for i in range(0, len(grouped_dataset), batch_size):
        yield grouped_dataset[i : i + batch_size]["text"]

model = models.BPE(unk_token="<unk>")
path = Path(f"tokenizers/{type(model).__name__}/{VOCAB_SIZE}/vocab.json")
if path.exists() and not OVERWRITE:
    tokenizer = tokenizers.Tokenizer.from_file(path.as_posix())
else:
    pre_tokenizer = pre_tokenizers.Split(" ", behavior="removed")
    tokenizer = tokenizers.Tokenizer(model)
    tokenizer.pre_tokenizer = pre_tokenizer
    trainer = trainers.BpeTrainer(vocab_size=VOCAB_SIZE)
    tokenizer.train_from_iterator(tokenization_generator(), trainer, length=len(dataset))
    path.parent.mkdir(parents=True, exist_ok=True)
    tokenizer.save(path.as_posix())
```

```python
tokenizer = transformers.PreTrainedTokenizerFast(
    tokenizer_object=tokenizer,
    model_max_length=MAX_SIZE,
    
)
tokenizer.add_special_tokens(SPECIALS)
tokenizer._tokenizer.enable_padding()
tokenizer._tokenizer.enable_truncation(max_length=MAX_SIZE)
```

```python
def preprocess_function(examples):
    return tokenizer(examples["text"], truncation=True, max_length=MAX_SIZE)

path = Path(f"datasets/tokenized/{VOCAB_SIZE}")
if path.exists() and not OVERWRITE:
    tokenized_dataset = Dataset.load_from_disk(path)
else:
    tokenized_dataset = dataset.map(preprocess_function, batched=True)
    path.parent.mkdir(parents=True, exist_ok=True)
    tokenized_dataset.save_to_disk(path.as_posix())
```

```python
tokenized_dataset = dataset.map(preprocess_function, batched=True)
```

```python
tokenized_dataset = tokenized_dataset.train_test_split(test_size=0.1)
```

```python
from transformers import DataCollatorWithPadding, LongformerForSequenceClassification

torch.cuda.empty_cache()

print_gpu_utilization()

config = transformers.LongformerConfig(
    max_position_embeddings=MAX_SIZE,
    vocab_size=len(tokenizer),
    sep_token_id=tokenizer.sep_token_id,
    pad_token_id=tokenizer.pad_token_id,
    bos_token_id=tokenizer.bos_token_id,
    eos_token_id=tokenizer.eos_token_id,
    num_labels=2,
    hidden_size=768 // DOWNSCALE,
    num_hidden_layers=12 // DOWNSCALE,
    num_attention_heads=12 // DOWNSCALE,
    intermediate_size=3072 // DOWNSCALE,
)
model = LongformerForSequenceClassification(config)
data_collator = DataCollatorWithPadding(tokenizer=tokenizer)

print_gpu_utilization()
```

```python
model.to("cuda")
print_gpu_utilization()
```

```python
training_args = transformers.TrainingArguments(
    output_dir="models/longformer-clf",
    evaluation_strategy="epoch",
    save_strategy="epoch",
    learning_rate=2e-5,
    num_train_epochs=50,
    per_device_train_batch_size=1,
    per_device_eval_batch_size=1,
    weight_decay=0.01,
    load_best_model_at_end=True,
    # group_by_length=True,
    gradient_accumulation_steps=16,
)

trainer = transformers.Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset["train"],
    eval_dataset=tokenized_dataset["test"],
    data_collator=data_collator,
    tokenizer=tokenizer,
    callbacks=[transformers.EarlyStoppingCallback(early_stopping_patience=5)],
)

trainer.train()
```
